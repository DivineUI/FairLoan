```python
from google.colab import userdata
from groq import Groq

api_key = userdata.get('FairLoan_API_Key') # Pulling the key from secrets manager
client = Groq(api_key=api_key) # Creating a client object authenticated with key

response = client.chat.completions.create(model="llama-3.1-8b-instant",messages=[
    {"role": "user", "content": "In one sentence, explain why credit scores matter for loans."}])

print(response.choices[0].message.content)
```

```python
# Explanation-generation function
def generate_explanation(decision, income, credit_score, loan_amount, prior_default):
  #The prompt
    prompt = f"""You are helping explain a loan decision to an applicant in plain, simple language.
    Decision: {decision} Applicant's income: GHS{income} Applicant's credit score: {credit_score}
Loan amount requested: GHS{loan_amount} Prior loan default on file: {"Yes" if prior_default else "No"}
Write a short, clear, respectful explanation (3-4 sentences) of this decision for the applicant.
Base your explanation only on the information given above. Do not invent or assume any information not provided.
Do not state or imply any specific lending policy, threshold, or rule beyond the data provided."""

    response = client.chat.completions.create(model="llama-3.1-8b-instant",
                                              messages=[{"role": "user", "content": prompt}])

    return response.choices[0].message.content
```

```python
# Testing
# 1st sample(Approved)
result = generate_explanation(decision="Approved", income=75000, credit_score=720,
                              loan_amount=10000, prior_default=False )
print(result)

# 2nd sample(Not approved)
print("")
result = generate_explanation(decision="Not Approved",income=32000, credit_score=580,
                              loan_amount=15000, prior_default=False)
print(result)

# 3rd sample(Review)
print("")
result = generate_explanation(decision="Manual Review needed", income=50000, credit_score=650,
                              loan_amount=20000, prior_default=True)
print(result)
```
